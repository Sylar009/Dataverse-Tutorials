```python
# Calling a Function
def welcome_message():
    print("Welcome to Dataverse Tutorials")

welcome_message()
```

```python
# Function Arguments – Passing Data to Functions
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(16))
print(check_even_odd(7))
```

```python
# Types of Function Arguments
# 1. Default Arguments
def course_details(course, duration=3):
    print(course)
    print(duration)

course_details("Python")

# 2. Keyword Arguments
def student_info(name, age):
    print(name, age)

student_info(age=22, name="Amit")

# 3. Positional Arguments
def display(name, age):
    print(name, age)

display("Ravi", 25)

# 4. Arbitrary Arguments
def show_data(*values, **details):
    for v in values:
        print(v)

    for k, d in details.items():
        print(k, d)

show_data("Python", "ML", level="Beginner", duration=30)
```

```python
# Nested Functions
def outer_function():
    message = "Learning Python Functions"

    def inner_function():
        print(message)

    inner_function()

outer_function()
```

```python
# Anonymous Functions – Lambda
square = lambda x: x * x
print(square(5))
```

```python
# Return Statement
def calculate_square(num):
    return num ** 2

print(calculate_square(4))
```

```python
# Pass by Reference vs Pass by Value
def modify_list(data):
    data[0] = 99

values = [1, 2, 3]
modify_list(values)
print(values)
```

```python
# Recursive Functions
def factorial(num):
    if num == 0:
        return 1
    return num * factorial(num - 1)

print(factorial(4))
```
