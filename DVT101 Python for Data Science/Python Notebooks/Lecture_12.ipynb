```python
# Local Variable Example – Inside Function
def show_message():
    text_inside = "Hello from function scope"
    print(text_inside)

show_message()
```

```python
# Why Local Variables Cannot Be Used Outside
def greet_user():
    welcome_text = "Welcome!"
    print("Inside:", welcome_text)

greet_user()
print("Outside:", welcome_text)
```

```python
# Using Global Variable Inside Function
app_message = "Python is powerful"

def display_message():
    print("Inside function:", app_message)

display_message()
print("Outside function:", app_message)
```

```python
# Local Variable Shadows Global Variable
status = "Global Status"

def check_status():
    status = "Local Status"
    print(status)

check_status()
print(status)
```

```python
# Why Global Cannot Be Modified Directly
count = 10

def update_count():
    count += 5
    print(count)

update_count()
```

```python
# Using global Keyword Safely
score = 50

def increase_score():
    global score
    score += 10
    print(score)

increase_score()
print(score)
```
