```python
# *args Example
def calculate_sum(*numbers):
    total = 0
    for value in numbers:
        total = total + value
    return total

print(calculate_sum(10, 20, 30, 40))
```

```python
# **kwargs Introduction
def show_profile(**details):
    for key, value in details.items():
        print(key, ":", value)

show_profile(name="Rahul", age=28, city="Delhi")
```

```python
# Combining *args and **kwargs
def enroll_student(*subjects, **info):
    print("Subjects:", subjects)
    print("Student Info:", info)

enroll_student("Math", "Science", name="Anita", age=21)
```
