```python
# Example 1: Car Class
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def show_details(self):
        return self.brand, self.model

car1 = Car("Toyota", "Corolla")
print(car1.show_details())
```

```python
# Example 2: Object Initialization
class TopicHolder:
    def __init__(self, topic):
        self.topic_name = topic

    def show_topic(self):
        print("Topic:", self.topic_name)

obj = TopicHolder("Python")
obj.show_topic()
```

```python
# Example 3: Circle Area Calculation
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius ** 2

c1 = Circle(5)
print("Area:", c1.calculate_area())
```
