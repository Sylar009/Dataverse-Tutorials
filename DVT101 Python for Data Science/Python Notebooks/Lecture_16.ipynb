```python
# Assigning Function to Variable
def greet_user(username):
    return f"Welcome, {username}!"

greeting = greet_user

print(greeting("Emma"))
```

```python
# Returning Function from Function
def message_creator(text):
    def inner_message():
        return f"Message received: {text}"
    return inner_message

result_func = message_creator("Hello World")
print(result_func())
```

```python
# Passing Function as Argument
def welcome(name):
    return f"Hello {name}"

def execute_function(func, value):
    return func(value)

print(execute_function(welcome, "Alex"))
```

```python
# Storing Functions in Data Structures
def add_numbers(a, b):
    return a + b

def subtract_numbers(a, b):
    return a - b

operations = {
    "add": add_numbers,
    "subtract": subtract_numbers
}

print(operations["add"](10, 5))
print(operations["subtract"](10, 5))
```
