```python
# Lambda Uppercase Example
text = "Dataverse"
to_upper = lambda value: value.upper()
print(to_upper(text))
```

```python
# Conditional Lambda
check_number = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(check_number(10))
print(check_number(-4))
print(check_number(0))
```

```python
# map() with Lambda
nums = [1, 2, 3, 4]
doubled = map(lambda x: x * 2, nums)
print(list(doubled))
```
