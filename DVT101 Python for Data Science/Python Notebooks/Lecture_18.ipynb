```python
# Example 1 – String to Integer Conversion
values = ['1', '2', '3', '4']
result = map(int, values)
print(list(result))
```

```python
# map() with Custom Function
def double_value(num):
    return num * 2

numbers = [1, 2, 3, 4]
output = list(map(double_value, numbers))
print(output)
```

```python
# map() with lambda (Most Common Use)
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x ** 2, nums))
print(squares)
```

```python
# map() with Multiple Iterables
a = [1, 2, 3]
b = [4, 5, 6]
res = map(lambda x, y: x + y, a, b)
print(list(res))
```

```python
# Text Processing – Uppercase Conversion
fruits = ['apple', 'banana', 'cherry']
result = list(map(str.upper, fruits))
print(result)
```

```python
# Data Cleaning – Remove Whitespaces
texts = ['  hello  ', '  world ', ' python  ']
cleaned = list(map(str.strip, texts))
print(cleaned)
```

```python
# Temperature Conversion (Real-World Example)
celsius = [0, 20, 37, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```
