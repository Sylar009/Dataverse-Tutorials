```python
# String Concatenation using reduce()
from functools import reduce

words = ["Python", "is", "powerful"]
sentence = reduce(lambda a, b: a + " " + b, words)
print(sentence)
```

```python
# reduce() with Named Function (Sum)
from functools import reduce

def add(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
res = reduce(add, a)
print(res)
```

```python
# Factorial using Lambda
from functools import reduce
a = [1, 2, 3, 4, 5]
res = reduce(lambda x, y: x * y, a)
print(res)
```

```python
# reduce() with operator module
import functools
import operator

a = [1, 3, 5, 6, 2]

print(functools.reduce(operator.add, a))
print(functools.reduce(operator.mul, a))
```

```python
# Using Initializer
from functools import reduce
a = [1, 2, 3]
res = reduce(lambda x, y: x + y, a, 10)
print(res)
```

```python
# accumulate() Example
from itertools import accumulate
from operator import add

a = [1, 2, 3, 4, 5]
res = accumulate(a, add)
print(list(res))
```
