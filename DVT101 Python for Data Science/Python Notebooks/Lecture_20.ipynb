```python
# Basic Example (Words starting with “a”)
def starts_a(w):
    return w.startswith("a")

li = ["apple", "banana", "avocado", "cherry", "apricot"]
res = filter(starts_a, li)
print(list(res))
```

```python
# filter() with Named Function (Even Numbers)
def even(n):
    return n % 2 == 0

a = [1, 2, 3, 4, 5, 6]
b = filter(even, a)
print(list(b))
```

```python
# filter() with Lambda Function
a = [1, 2, 3, 4, 5, 6]
b = filter(lambda x: x % 2 == 0, a)
print(list(b))
```

```python
# filter() + map() (Pipeline Example)
a = [1, 2, 3, 4, 5, 6]
b = filter(lambda x: x % 2 == 0, a)
c = map(lambda x: x * 2, b)
print(list(c))
```

```python
# Filtering Strings by Length
a = ["apple", "banana", "cherry", "kiwi", "grape"]
b = filter(lambda w: len(w) > 5, a)
print(list(b))
```

```python
# filter(None, iterable)
L = ["apple", "", None, "banana", 0, "cherry"]
A = filter(None, L)
print(list(A))
```
