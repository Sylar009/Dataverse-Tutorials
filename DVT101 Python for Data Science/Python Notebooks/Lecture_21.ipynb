```python
# Simple Inner Function Example
def fun1(msg):
    def fun2():
        print(msg)
    fun2()

fun1("Hello")
```

```python
# Modifying Outer Variable using nonlocal
def fun1():
    a = 45
    def fun2():
        nonlocal a
        a = 54
        print(a)
    fun2()
    print(a)

fun1()
```

```python
# Closure Example
def fun1(a):
    def fun2():
        print(a)
    return fun2

closure_func = fun1("Hello, Closure!")
closure_func()
```

```python
# Real World Use: Data Cleaning
def process_data(data):
    def clean_data():
        return [item.strip() for item in data]
    return clean_data()

print(process_data(["  Python  ", "  Inner Function  "]))
```

```python
# Logging Example (Decorator Style)
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Executing {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def add(a, b):
    return a + b

print(add(3, 4))
```
