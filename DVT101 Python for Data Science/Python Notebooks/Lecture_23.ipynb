```python
# Creating Strings
text1 = 'Python'
text2 = "Python"
print(text1)
print(text2)
```

```python
# Multi-Line Strings
message = """Welcome to Python
This is a multi-line string
Used for long text"""
print(message)
```

```python
# Accessing Characters
word = "Dataverse"
print(word[0])
print(word[4])
print(word[-1])
```

```python
# String Slicing
text = "Programming"
print(text[0:6])
print(text[3:])
print(text[::-1])
```

```python
# Iterating Over String
name = "Python"
for ch in name:
    print(ch)
```

```python
# String Immutability
word = "python"
word = "P" + word[1:]
print(word)
```

```python
# Updating & Deleting Strings
msg = "hello world"
updated = msg.replace("world", "Python")
print(updated)
```

```python
# Common String Methods
data = "  Python Rocks  "
print(len(data))
print(data.strip())
print(data.upper())
```

```python
# Concatenation & Repetition
greet = "Hi "
print(greet * 3)
```

```python
# String Formatting
user = "Aman"
score = 95
print(f"User {user} scored {score} marks")
```

```python
# Membership Testing
text = "Learn Python"
print("Python" in text)
print("Java" in text)
```
