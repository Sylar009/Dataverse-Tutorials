```python
# Creating Lists (Square Brackets)
a = [1, 2, 3, 4, 5]
b = ['apple', 'banana', 'cherry']
c = [1, 'hello', 3.14, True]

print(a)
print(b)
print(c)
```

```python
# list() Constructor
a = list((1, 2, 3, 'apple', 4.5))
b = list("DVT")
print(a)
print(b)
```

```python
# Repeating Lists
a = [2] * 5
b = [0] * 7
print(a)
print(b)
```

```python
# Indexing & Slicing
a = [10, 20, 30, 40, 50]
print(a[0])
print(a[-1])
print(a[1:4])
```

```python
# Adding Elements
a = []
a.append(10)
a.insert(0,5)
a.extend([15,20,25])
print(a)
```

```python
# Updating List
a = [10,20,30,40]
a[1] = 25
print(a)
```

```python
# Removing Elements
a = [10,20,30,40,50]
a.remove(30)
p = a.pop(1)
del a[0]
print(a)
```

```python
# Iteration
for item in ['apple','banana','cherry']:
    print(item)
```

```python
# Nested Lists
matrix = [[1,2,3],[4,5,6],[7,8,9]]
print(matrix[1][2])
```

```python
# List Comprehension
squares = [x**2 for x in range(1,6)]
print(squares)
```
