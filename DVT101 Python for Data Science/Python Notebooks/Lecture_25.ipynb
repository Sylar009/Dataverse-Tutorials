```python
# Creating Tuples – Basic Ways
t1 = ()
print(t1)

t2 = ("Python", "R")
print(t2)

lst = [10, 20, 30]
print(tuple(lst))

t3 = tuple("Code")
print(t3)
```

```python
# Mixed & Nested Tuples
data = (10, "Python", 3.5, True)
print(data)

nums = (1, 2, 3)
langs = ("Java", "C++")
combo = (nums, langs)
print(combo)

repeat = ("AI",) * 3
print(repeat)

word = ("Code")
count = 4
for i in range(count):
    word = (word,)
    print(word)
```

```python
#  Accessing & Unpacking Tuples
letters = tuple("Python")
print(letters[0])

print(letters[1:4])
print(letters[:3])

values = ("Red", "Green", "Blue")
x, y, z = values
print(x)
print(y)
print(z)
```

```python
# Tuple Concatenation
a = (1, 2, 3)
b = ("X", "Y", "Z")

c = a + b
print(c)
```

```python
# Tuple Slicing
chars = tuple("PROGRAMMING")

print(chars[2:])
print(chars[::-1])
print(chars[3:8])
```

```python
# Deleting Tuple
items = (100, 200, 300)
del items

print(items)
items = (100, 200, 300)
del items

print(items)
```

```python
# Tuple Unpacking with *
nums = (5, 10, 15, 20, 25)

first, *middle, last = nums

print(first)
print(middle)
print(last)
```
