```python
# Basic Dictionary Example
info = {"name": "Jake", "age": 22}
print(info)
```

```python
# Creating Dictionary
data1 = {1: "Alpha", 2: "Beta", 3: "Gamma"}
print(data1)

data2 = dict(a="Red", b="Blue", c="Green")
print(data2)
```

```python
# Accessing Dictionary
record = {"user": "Kat", 1: "Python", (1, 2): [1, 2, 4]}
print(record["user"])
print(record.get("user"))
```

```python
# Add & Update
details = {1: "Alpha", 2: "Beta", 3: "Gamma"}
details["age"] = 22
details[1] = "Python dict"
print(details)
```

```python
# Remove Items
data = {1: "Alpha", 2: "Beta", 3: "Gamma", "age": 22}
del data["age"]
print(data)

val = data.pop(1)
print(val)

key, val = data.popitem()
print(f"Key: {key}, Value: {val}")

data.clear()
print(data)
```

```python
# Iteration
info = {1: "Alpha", 2: "Beta", "age": 22}

for key in info:
    print(key)

for value in info.values():
    print(value)

for key, value in info.items():
    print(f"{key}: {value}")
```

```python
# Nested Dictionary
profile = {
    1: "Alpha",
    2: "Beta",
    3: {"A": "Welcome", "B": "To", "C": "Python"}
}
print(profile)
```
