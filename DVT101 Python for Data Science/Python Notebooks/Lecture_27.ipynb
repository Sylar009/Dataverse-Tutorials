```python
# Creating a Set
nums = {1, 2, 3, 4}
print(nums)
```

```python
# Creating using set()
letters = set("DataverseTutorials")
print(letters)

items = set(["DVT", "For", "Python"])
print(items)
```

```python
# Unordered & Unindexed
nums = {3, 1, 4, 1, 5, 9, 2}
print(nums)

try:
    print(nums[0])
except TypeError as err:
    print(err)
```

```python
# Adding Elements
data = {1, 2, 3}
data.add(4)
data.update([5, 6])
print(data)
```

```python
# Accessing Elements
tags = set(["DVT", "For", "Python."])

for item in tags:
    print(item, end=" ")

print("DVT" in tags)
```

```python
# remove() & discard()
nums = {1, 2, 3, 4, 5}
nums.remove(3)

try:
    nums.remove(10)
except KeyError as e:
    print("Error:", e)

nums.discard(4)
nums.discard(10)
print(nums)
```

```python
# pop() & clear()
nums = {1, 2, 3, 4, 5}
val = nums.pop()
print(val)
print(nums)

nums.clear()
print(nums)
```

```python
# Frozen Set (Immutable Set)
frozen = frozenset([1, 2, 3, 4, 5])
print(frozen)
```

```python
# Typecasting Objects into Set
nums = [1, 2, 3, 3, 4, 5, 5, 6]
unique_nums = set(nums)
print(unique_nums)

text = "DataverseTutorials"
letters = set(text)
print(letters)

info = {1: "One", 2: "Two", 3: "Three"}
keys_set = set(info)
print(keys_set)
```
