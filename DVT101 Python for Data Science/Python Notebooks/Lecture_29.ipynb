```python
# For Loop vs List Comprehension
nums_box = [1, 2, 3, 4, 5]
doubled_box = []

for value in nums_box:
    doubled_box.append(value * 2)

print(doubled_box)
```

```python
# Simple Transformation
nums_box = [2, 3, 4, 5]
squared_box = [item ** 2 for item in nums_box]
print(squared_box)
```

```python
# Conditional Filtering
nums_box = [1, 2, 3, 4, 5, 6]
even_box = [item for item in nums_box if item % 2 == 0]
print(even_box)
```

```python
# If–Else in List Comprehension
marks = [35, 60, 75, 20]
status = ["Pass" if m >= 40 else "Fail" for m in marks]
print(status)
```

```python
# Using Range
series = [num for num in range(10)]
print(series)
```

```python
# Nested List Comprehension
pairs = [(x, y) for x in range(3) for y in range(3)]
print(pairs)
```

```python
# Flattening Nested Lists
matrix = [[1, 2], [3, 4], [5, 6]]
flat_list = [val for row in matrix for val in row]
print(flat_list)
```
