```python
#Taking Input in Python
user_name = input("Enter your name: ")
print("Hello", user_name, "Welcome!")
```

```python
print("Hello, World!")
```

```python
# Printing Variables
actor_name = "Chris"
print(actor_name)

actor_name = "Emma"
actor_age = 28
actor_city = "London"
print(actor_name, actor_age, actor_city)
```

```python
# Taking Multiple Inputs using split()
boys_count, girls_count = input("Enter boys and girls count: ").split()
print("Boys:", boys_count)
print("Girls:", girls_count)
```

```python
# Typecasting Input in Python
rose_count = int(input("How many roses?: "))
print(rose_count)
```

```python
rose_price = float(input("Enter price per rose: "))
print(rose_price)
```

```python
# Finding Data Type in Python
text_val = "Python"
num_val = 50
decimal_val = 12.75
tuple_val = ("Data", "Science")
list_val = ["ML", "AI"]
dict_val = {"course": "Python", "level": "Beginner"}

print(type(text_val))
print(type(num_val))
print(type(decimal_val))
print(type(tuple_val))
print(type(list_val))
print(type(dict_val))
```
