```python
# Creating Counter
from collections import Counter

data_stream = [1, 2, 2, 3, 3, 3]
count_map = Counter(data_stream)

print(count_map)
```

```python
# Accessing Counter Values
from collections import Counter

score_log = Counter([10, 20, 20, 30, 30, 30])

print(score_log[10])
print(score_log[20])
print(score_log[40])
```

```python
# Updating Counter Data
from collections import Counter

event_log = Counter([1, 2, 2])
event_log.update([2, 3, 3, 3])

print(event_log)
```

```python
# elements() Method
from collections import Counter

bag = Counter([1, 2, 2, 3, 3, 3])
expanded = list(bag.elements())

print(expanded)
```

```python
# most_common()
from collections import Counter

votes = Counter([1, 2, 2, 3, 3, 3])
top_two = votes.most_common(2)

print(top_two)
```

```python
# Incrementing Counts
from collections import Counter

tracker = Counter([5, 6, 6])
tracker[6] += 1
tracker[7] += 1

print(tracker)
```

```python
# subtract()
from collections import Counter

inventory = Counter([1, 2, 2, 3, 3, 3])
inventory.subtract([2, 3, 3])

print(inventory)
```

```python
# Arithmetic Operations
from collections import Counter

a_counts = Counter([1, 2, 2, 3])
b_counts = Counter([2, 3, 3, 4])

print(a_counts + b_counts)
print(a_counts - b_counts)
print(a_counts & b_counts)
print(a_counts | b_counts)
```
