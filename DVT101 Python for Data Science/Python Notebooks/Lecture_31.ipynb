```python
# Creating a Heap (heapify)
import heapq
nums_heap = [25, 20, 15, 30, 40]
heapq.heapify(nums_heap)
print(nums_heap)
```

```python
# Using Heap as a Max-Heap
import heapq
values = [10, 20, 15, 30, 40]
neg_heap = [-n for n in values]
heapq.heapify(neg_heap)
print(-neg_heap[0])
```

```python
# heappush & heappop
import heapq
task_heap = [10, 20, 15, 30, 40]
heapq.heapify(task_heap)
heapq.heappush(task_heap, 5)
small = heapq.heappop(task_heap)
print(small)
print(task_heap)
```

```python
# heappushpop
import heapq
prio = [10, 20, 15, 30, 40]
heapq.heapify(prio)
res = heapq.heappushpop(prio, 5)
print(res)
print(prio)
```

```python
# nlargest & nsmallest
import heapq
data = [10, 20, 15, 30, 40]
heapq.heapify(data)
top_big = heapq.nlargest(3, data)
top_small = heapq.nsmallest(3, data)
print(top_big)
print(top_small)
```

```python
# heapreplace
import heapq
h = [10, 20, 15, 30, 40]
heapq.heapify(h)
old = heapq.heapreplace(h, 5)
print(old)
print(h)
```

```python
# merge
import heapq
h1 = [10, 20, 30]
h2 = [5, 15, 25]
merged = list(heapq.merge(h1, h2))
print(merged)
```
