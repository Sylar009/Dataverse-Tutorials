```python
# Creating a Deque
from collections import deque

deq_box = deque(['name', 'age', 'DOB'])
print(deq_box)
```

```python
# Adding Elements
from collections import deque

dq = deque([10, 20, 30])
dq.append(40)
dq.appendleft(5)
dq.extend([50, 60])
dq.extendleft([0, 2])
print(dq)
```

```python
# Removing Elements
from collections import deque

dq = deque([5, 10, 20, 30, 40])
dq.remove(20)
dq.pop()
dq.popleft()
print(dq)
```

```python
# Indexing & Length
import collections

dq = collections.deque([1, 2, 3, 4])
print(dq[0])
print(dq[-1])
print(len(dq))
```

```python
# Count, Rotate, Reverse
from collections import deque

dq = deque([10, 20, 30, 40, 20, 30, 20])
print(dq.count(20))
dq.rotate(2)
print(dq)
dq.rotate(-3)
print(dq)
dq.reverse()
print(dq)
```
