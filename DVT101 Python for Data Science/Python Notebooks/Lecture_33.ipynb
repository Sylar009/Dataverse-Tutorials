```python
# Basic OrderedDict Creation
from collections import OrderedDict

fruit_box = OrderedDict()
fruit_box['apple'] = 1
fruit_box['banana'] = 2
fruit_box['cherry'] = 3

print(list(fruit_box.items()))
```

```python
# dict vs OrderedDict
from collections import OrderedDict

normal_map = {}
normal_map['a'] = 1
normal_map['b'] = 2
normal_map['c'] = 3
normal_map['d'] = 4

for key, val in normal_map.items():
    print(key, val)

order_map = OrderedDict()
order_map['d'] = 4
order_map['b'] = 2
order_map['a'] = 1
order_map['c'] = 3

for key, val in order_map.items():
    print(key, val)
```

```python
# Updating Value Does Not Change Order
from collections import OrderedDict

rank_map = OrderedDict([('a', 1), ('b', 2), ('c', 3), ('d', 4)])
rank_map['c'] = 5

for k, v in rank_map.items():
    print(k, v)
```

```python
# Equality Check
from collections import OrderedDict

od1 = OrderedDict([('a', 1), ('b', 2), ('c', 3)])
od2 = OrderedDict([('c', 3), ('b', 2), ('a', 1)])

print(od1 == od2)
```

```python
# Reversing OrderedDict
from collections import OrderedDict

base = OrderedDict([('a', 1), ('b', 2), ('c', 3)])
rev = OrderedDict(reversed(list(base.items())))

for k, v in rev.items():
    print(k, v)
```

```python
# popitem from Ends
from collections import OrderedDict

d = OrderedDict([('a', 1), ('b', 2), ('c', 3)])
res = d.popitem(last=True)
print(res)
```

```python
# move_to_end()
from collections import OrderedDict

track = OrderedDict([('a', 1), ('b', 2), ('c', 3)])
track.move_to_end('a')
track.move_to_end('b', last=False)

for k, v in track.items():
    print(k, v)
```

```python
# Delete & Reinsert
from collections import OrderedDict

od = OrderedDict([('a', 1), ('b', 2), ('c', 3), ('d', 4)])
od.pop('c')

for k, v in od.items():
    print(k, v)

od['c'] = 3

for k, v in od.items():
    print(k, v)
```
