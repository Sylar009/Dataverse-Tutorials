```python
# Basic Creation & Missing Key
from collections import defaultdict

box = defaultdict(list)

box['fruits'].append('apple')
box['vegetables'].append('carrot')

print(box)
print(box['juices'])
```

```python
# list as Default Factory
from collections import defaultdict

store = defaultdict(list)

for num in range(5):
    store[num].append(num)

print(store)
```

```python
# int as Default Factory
from collections import defaultdict

counter = defaultdict(int)
data = [1, 2, 3, 4, 2, 4, 1, 2]

for val in data:
    counter[val] += 1

print(counter)
```

```python
# str as Default Factory
from collections import defaultdict

text_map = defaultdict(str)

text_map['greeting'] = 'Hello'
print(text_map)
```

```python
# Grouping Words
from collections import defaultdict

words = ["apple", "ant", "banana", "bat", "carrot", "cat"]
grouped = defaultdict(list)

for word in words:
    grouped[word[0]].append(word)

print(grouped)
```

```python
# missing Behavior
from collections import defaultdict

auto = defaultdict(lambda: "Not Present")
auto["a"] = 1
auto["b"] = 2

print(auto.__missing__('x'))
print(auto.__missing__('d'))
print(auto['a'])
```
