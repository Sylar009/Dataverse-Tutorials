```python
# Class, Attributes & Constructor
class Animal:
    category = "Mammal"

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
# Object Creation & Access
pet1 = Animal("Buddy", 3)

print(pet1.name)
print(pet1.category)
```

```python
# Inheritance
class Vehicle:
    def move(self):
        print("Vehicle is moving")

class Car(Vehicle):
    def honk(self):
        print("Car is honking")

car = Car()
car.move()
car.honk()
```

```python
# Polymorphism
class Dog:
    def speak(self):
        print("Dog barks")

class Cat:
    def speak(self):
        print("Cat meows")

dog = Dog()
dog.speak()

cat = Cat()
cat.speak()
```

```python
# Encapsulation
class Account:
    def __init__(self, balance):
        self.__balance = balance
```
