```python
# Class Attribute
class Dog:
    sound = "bark"

dog1 = Dog()
print(dog1.sound)
```

```python
# Constructor & Object Creation
class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.species)
```

```python
# str Method
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old."

dog1 = Dog("Buddy", 3)
dog2 = Dog("Charlie", 5)

print(dog1)
print(dog2)
```

```python
# Class vs Instance Variables
class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)
dog2 = Dog("Charlie", 5)

print(dog1.species)
print(dog1.name)
print(dog2.name)

dog1.name = "Max"
print(dog1.name)

Dog.species = "Feline"
print(dog1.species)
print(dog2.species)
```
