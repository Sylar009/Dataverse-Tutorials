```python
# Default Arguments (Overloading Mimic)
class Multiplier:
    def multiply(self, x=1, y=1, *rest):
        total = x * y
        for num in rest:
            total *= num
        return total

calc = Multiplier()
print(calc.multiply())
print(calc.multiply(4))
print(calc.multiply(2, 3))
print(calc.multiply(2, 3, 4))
```

```python
# Runtime Polymorphism (Method Overriding)
class Creature:
    def sound(self):
        return "Some sound"

class Dog(Creature):
    def sound(self):
        return "Bark"

class Cat(Creature):
    def sound(self):
        return "Meow"

animals = [Dog(), Cat(), Creature()]
for a in animals:
    print(a.sound())
```

```python
# Built-in Polymorphism
print(len("Hello"))
print(len([1, 2, 3]))

print(max(1, 3, 2))
print(max("a", "z", "m"))
```

```python
# Duck Typing
class Pen:
    def use(self):
        return "Writing"

class Eraser:
    def use(self):
        return "Erasing"

def perform_task(tool):
    print(tool.use())

perform_task(Pen())
perform_task(Eraser())
```

```python
# Operator Polymorphism
print(5 + 10)
print("Hello " + "World!")
print([1, 2] + [3, 4])
```
