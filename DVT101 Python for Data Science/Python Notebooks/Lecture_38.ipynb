```python
# Simple Parent–Child Class Concept
class Creature:
    def __init__(self, label):
        self.label = label

    def show_info(self):
        print("Creature label:", self.label)

class Tiger(Creature):
    def roar(self):
        print(self.label, "roars")

t = Tiger("Sheru")
t.show_info()
t.roar()
```

```python
# super() Function Concept
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def info(self):
        print("Brand:", self.brand)

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def details(self):
        print(self.brand, "model is", self.model)

c = Car("Tesla", "Model S")
c.info()
c.details()
```

```python
# Single Inheritance
class User:
    def __init__(self, username):
        self.username = username

class Admin(User):
    def show_role(self):
        print(self.username, "is an admin")

a = Admin("Rohit")
print("Name:", a.username)
a.show_role()
```

```python
# Multiple Inheritance
class Writer:
    def __init__(self, pen_name):
        self.pen_name = pen_name

class Speaker:
    def __init__(self, language):
        self.language = language

class Influencer(Writer, Speaker):
    def __init__(self, pen_name, language):
        Writer.__init__(self, pen_name)
        Speaker.__init__(self, language)

    def profile(self):
        print(self.pen_name, "speaks in", self.language)

i = Influencer("Nova", "English")
i.profile()
```

```python
# Multilevel Inheritance
class Device:
    def __init__(self, brand):
        self.brand = brand

class Mobile(Device):
    def feature(self):
        print(self.brand, "is a mobile device")

class SmartPhone(Mobile):
    def camera(self):
        print(self.brand, "has a smart camera")

s = SmartPhone("Pixel")
s.feature()
s.camera()
```

```python
# Hybrid Inheritance
class Human:
    def __init__(self, name):
        self.name = name

class Engineer(Human):
    def role(self):
        print(self.name, "is an engineer")

class Project:
    def __init__(self, title):
        self.title = title

class Lead(Engineer, Project):
    def __init__(self, name, title):
        Engineer.__init__(self, name)
        Project.__init__(self, title)

    def details(self):
        print(self.name, "leads project", self.title)

l = Lead("Aman", "AI System")
l.role()
l.details()
```
