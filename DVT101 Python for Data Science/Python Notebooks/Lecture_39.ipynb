```python
# Abstract Class Instantiation Rule
from abc import ABC, abstractmethod

class ShapeBase(ABC):
    @abstractmethod
    def draw(self):
        pass

s = ShapeBase()
```

```python
# Simple ABC Example
from abc import ABC, abstractmethod

class GreeterBase(ABC):
    @abstractmethod
    def greet(self):
        pass

class HindiGreeter(GreeterBase):
    def greet(self):
        return "Namaste!"

h = HindiGreeter()
print(h.greet())
```

```python
# Abstract vs Concrete
from abc import ABC, abstractmethod

class Machine(ABC):
    @abstractmethod
    def start(self):
        pass

    def stop(self):
        return "Machine stopped"

class Robot(Machine):
    def start(self):
        return "Robot started"

r = Robot()
print(r.start())
print(r.stop())
```

```python
# Abstract Properties
from abc import ABC, abstractmethod

class VehicleBase(ABC):
    @property
    @abstractmethod
    def type_name(self):
        pass

class Bike(VehicleBase):
    @property
    def type_name(self):
        return "Two Wheeler"

b = Bike()
print(b.type_name)
```
