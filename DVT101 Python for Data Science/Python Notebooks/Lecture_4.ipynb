```python
#Variable Assignment in Python
count = 5
user_name = "Samantha"

print(count)
print(user_name)
```

```python
# Dynamic Typing in Python
data = 10
data = "Now text value"
print(data)
```

```python
# Multiple Assignments in Python
x = y = z = 100
print(x, y, z)

num, rate, text = 1, 2.5, "Python"
print(num, rate, text)
```

```python
# Type Casting in Python
text_num = "10"
number = int(text_num)

items = 5
price = float(items)

age = 25
age_text = str(age)

print(number)
print(price)
print(age_text)
```

```python
# Getting Data Type using type()
a = 42
b = 3.14
c = "Hello"
d = [1, 2, 3]
e = {"key": "value"}
f = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
```

```python
# Swapping Variables
a, b = 5, 10
a, b = b, a
print(a, b)
```

```python
# String Length
word = "Python"
length = len(word)
print("Length of the word:", length)
```
