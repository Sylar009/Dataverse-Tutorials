```python
# Program: Public Members
class Staff:
    def __init__(self, full_name):
        self.full_name = full_name

    def show_name(self):
        print(self.full_name)

member = Staff("Amit Sharma")
member.show_name()
print(member.full_name)
```

```python
# Program: Protected Members
class Worker:
    def __init__(self, name, age):
        self.name = name
        self._age = age

class TeamMember(Worker):
    def display_age(self):
        print("Age is", self._age)

tm = TeamMember("Riya", 28)
tm.display_age()
```

```python
# Program: Private Members
class Payroll:
    def __init__(self, employee_name, salary):
        self.employee_name = employee_name
        self.__salary = salary

    def show_salary(self):
        print("Salary is", self.__salary)

pay = Payroll("Neha", 70000)
pay.show_salary()
```

```python
# Program: Protected vs Private Methods
class Account:
    def __init__(self):
        self.balance = 1000

    def _show_balance(self):
        print("Balance:", self.balance)

    def __update_balance(self, amount):
        self.balance += amount

    def deposit(self, amount):
        if amount > 0:
            self.__update_balance(amount)
            self._show_balance()

acc = Account()
acc._show_balance()
acc.deposit(500)
```

```python
# Program: Getter and Setter Methods
class SalaryBox:
    def __init__(self):
        self.__pay = 45000

    def get_pay(self):
        return self.__pay

    def set_pay(self, value):
        if value > 0:
            self.__pay = value

sb = SalaryBox()
print(sb.get_pay())

sb.set_pay(60000)
print(sb.get_pay())
```
