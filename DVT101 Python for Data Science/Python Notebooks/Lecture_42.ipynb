```python
# Program: Catching Specific Exceptions
try:
    value = int("abc")
    result = 10 / value
except ValueError:
    print("Invalid number provided")
except ZeroDivisionError:
    print("Division by zero not allowed")
```

```python
# Program: Multiple Except Blocks
try:
    data = [10, 20]
    total = data[0] + data[5]
except IndexError:
    print("Index does not exist")
except TypeError:
    print("Invalid data type used")
```

```python
# Program: Catching Multiple Exceptions Together
try:
    number = int("ten")
    output = 100 / number
except (ValueError, ZeroDivisionError):
    print("Invalid calculation")
```

```python
# Program: Catch-All Exception Handler
try:
    result = "50" + 20
except ArithmeticError:
    print("Math related error")
except:
    print("Unexpected error occurred")
```

```python
# Program: Raising Exceptions Manually
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    print("Age set to", age)

try:
    set_age(-3)
except ValueError as err:
    print(err)
```

```python
# Program: Custom Exceptions
class BalanceError(Exception):
    pass

def withdraw(amount):
    if amount > 5000:
        raise BalanceError("Withdrawal limit exceeded")
    print("Withdrawal successful")

try:
    withdraw(7000)
except BalanceError as e:
    print(e)
```
