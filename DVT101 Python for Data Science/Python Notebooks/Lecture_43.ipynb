```python
# AttributeError
class Student:
    pass

stud = Student()
try:
    print(stud.roll_number)
except AttributeError as err:
    print(err)
```

```python
# IndexError
numbers = [5, 10, 15]

try:
    print(numbers[5])
except IndexError as err:
    print(err)
```

```python
# KeyError
profile = {"name": "Amit", "age": 25}

try:
    print(profile["salary"])
except KeyError as err:
    print(err)
```

```python
# OSError / FileNotFoundError
try:
    file = open("data_file.txt")
except FileNotFoundError as err:
    print(err)
```

```python
# StopIteration
items = [1, 2]
iterator = iter(items)

try:
    print(next(iterator))
    print(next(iterator))
    print(next(iterator))
except StopIteration:
    print("No more elements left")
```

```python
# KeyboardInterrupt
try:
    while True:
        print("Running...")
except KeyboardInterrupt:

    print("Program interrupted by user")
```
