```python
# Program: Enhanced Custom Exception (with Error Code)
class AgeLimitError(Exception):
    def __init__(self, age_value,
                 message="Age must be between 1 and 100",
                 code=9001):
        self.age_value = age_value
        self.message = message
        self.code = code
        super().__init__(self.message)

    def __str__(self):
        return f"[Code {self.code}] {self.age_value} -> {self.message}"
try:
    raise AgeLimitError(150)
except AgeLimitError as err:
    print(err)
```

```python
# Program: Using Standard Exception as Base (ValueError)
class InvalidScoreError(ValueError):
    def __init__(self, score):
        self.score = score
        super().__init__(f"Invalid score: {score}")

try:
    raise InvalidScoreError(-20)
except InvalidScoreError as err:
    print(err)
```

```python
# Program: Custom Exception from RuntimeError
class ServiceFailureError(RuntimeError):
    def __init__(self, reason):
        self.reason = reason
        super().__init__(reason)

try:
    raise ServiceFailureError("Service temporarily unavailable")
except ServiceFailureError as err:
    print(err)
```

```python
# Program: Real-world Input Validation (Email)
class EmailFormatError(Exception):
    def __init__(self, email_value):
        self.email_value = email_value
        super().__init__("Invalid email format")

    def __str__(self):
        return f"{self.email_value} -> Invalid email format"

# Program: Real-world Validation Logic with Custom Exception
def register_email(user_email):
    if "@" not in user_email:
        raise EmailFormatError(user_email)
    print("Email registered:", user_email)

try:
    register_email("userexample.com")
except EmailFormatError as err:
    print(err)
```
