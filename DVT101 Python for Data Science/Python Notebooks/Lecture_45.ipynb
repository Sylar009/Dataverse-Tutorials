```python
# Program: Opening a File in Read Mode
data_file = open("sample_data.txt", "r")
print(data_file)
```

```python
# File Modes Explained (Read Context Program)
read_handle = open("sample_data.txt", "r")
status_check = read_handle.readable()
print(status_check)
read_handle.close()
```

```python
# File Object & Its Properties
file_ref = open("sample_data.txt", "r")
print(file_ref.name)
print(file_ref.mode)
print(file_ref.closed)
file_ref.close()
print(file_ref.closed)
```

```python
# Reading Data from Files
reader = open("sample_data.txt", "r")
file_text = reader.read()
print(file_text)
reader.close()
```

```python
# Program: Writing Data Using Safe Context
with open("output_data.txt", "w") as writer:
    writer.write("Welcome to Dataverse Tutorials\n")
    writer.write("Python file handling made simple")
print("Write operation completed")
```

```python
# Using the with Statement
with open("auto_close_demo.txt", "r") as file_handler:
    file_content = file_handler.read()
    print(file_content)
print("File handling block completed")
```

```python
# Exception Handling in File Operations
try:
    file_stream = open("exception_demo.txt", "r")
    data_inside = file_stream.read()
    print(data_inside)
finally:
    file_stream.close()
    print("File safely closed")
```
