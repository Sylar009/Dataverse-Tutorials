```python
# Reading Entire File Content
text_file = open("reading_demo.txt", "r")
full_text = text_file.read()
print(full_text)
text_file.close()
```

```python
# Reading a File Using with Statement
with open("reading_demo.txt", "r") as safe_reader:
    data_inside = safe_reader.read()
    print(data_inside)
```

```python
# Reading Files Line by Line (Loop Based)
line_file = open("reading_demo.txt", "r")
for single_line in line_file:
    print(single_line.strip())
line_file.close()
```

```python
# Reading File Using readline() Method
control_file = open("reading_demo.txt", "r")
current_line = control_file.readline()
while current_line:
    print(current_line.strip())
    current_line = control_file.readline()
control_file.close()
```

```python
# Reading Binary Files
binary_reader = open("binary_demo.bin", "rb")
binary_data = binary_reader.read()
print(binary_data)
binary_reader.close()
```

```python
# Reading Specific Portions of a File
preview_file = open("reading_demo.txt", "r")
snippet_text = preview_file.read(10)
print(snippet_text)
preview_file.close()
```

```python
# Reading CSV Files (Conceptual Program)
import csv
import io

csv_text = "Year,Product,Sales\n2022,Laptop,500\n2023,Tablet,300"
csv_buffer = io.StringIO(csv_text)

csv_reader = csv.reader(csv_buffer)
for row_data in csv_reader:
    print(row_data)
```

```python
# CSV Reading Backend Logic (Practical View)
import csv
import io

sales_data = "Item,Quantity\nPen,10\nBook,5"
memory_file = io.StringIO(sales_data)

reader_object = csv.reader(memory_file)
header = next(reader_object)

for record in reader_object:
    print("Item:", record[0], "Quantity:", record[1])
```
