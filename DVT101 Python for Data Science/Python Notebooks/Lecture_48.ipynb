```python
# Changing File Permissions
import os

target_file = "/content/Python_Workspace/secure_data.txt"
os.chmod(target_file, 0o600)
print("Permissions updated for:", target_file)
```

```python
# File Ownership Management (Unix/Linux)
import os

file_name = "/content/Python_Workspace/project_notes.txt"
user_id = 1001
group_id = 1001

os.chown(file_name, user_id, group_id)
print("Ownership updated for:", file_name)
```

```python
# File Metadata and Statistics
import os

info = os.stat("/content/Python_Workspace/project_notes.txt")

print("File size:", info.st_size)
print("Last modified time:", info.st_mtime)
print("Permission bits:", oct(info.st_mode)[-3:])
```

```python
# Error Handling in OS Operations
import os

try:
    with open("/content/Python_Workspace/missing_file.txt", "r") as reader:
        data = reader.read()
        print(data)
except FileNotFoundError:
    print("Error: File does not exist.")
except PermissionError:
    print("Error: Permission denied.")
```

```python
# File Existence and Size Checking
import os

file_check = "/content/Python_Workspace/project_notes.txt"

if os.path.exists(file_check):
    file_size = os.path.getsize(file_check)
    print("File exists with size:", file_size, "bytes")
else:
    print("File not found.")
```
