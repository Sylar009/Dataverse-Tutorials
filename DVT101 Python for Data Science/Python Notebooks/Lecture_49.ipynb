```python
# Creating and Inspecting a Path Object
from pathlib import Path

my_path = Path("projects/data/file.txt")
print(my_path)
```

```python
# Importing and Using Path Class
from pathlib import Path

base_dir = Path("projects")
full_path = base_dir / "reports" / "summary.txt"

print(full_path)
```

```python
# Directory Listing Concept
from pathlib import Path

folder = Path("projects")

for item in folder.iterdir():
    print(item)
```

```python
# Recursive File Searching
from pathlib import Path

root_dir = Path("projects")

for file_path in root_dir.rglob("*.py"):
    print(file_path)
```

```python
# Path Navigation Using Operator Overloading
from pathlib import Path

home = Path("projects")
config_file = home / "config" / "settings.json"

print(config_file)
```

```python
# Querying Path Properties
from pathlib import Path

file_path = Path("projects/reports/summary.txt")

print("Name:", file_path.name)
print("Extension:", file_path.suffix)
print("Parent:", file_path.parent)
```

```python
# File Reading and Writing Concept
from pathlib import Path

data_file = Path("projects/data/info.txt")

data_file.write_text("Welcome to Dataverse Tutorials")
content = data_file.read_text()

print(content)
```

```python
# PurePath Class Concept
from pathlib import PurePath

logical_path = PurePath("logs/2025/september/report.txt")
print(logical_path)
```

```python
# PurePosixPath vs PureWindowsPath
from pathlib import PurePosixPath, PureWindowsPath

linux_style = PurePosixPath("data/output/file.txt")
windows_style = PureWindowsPath("data/output/file.txt")

print(linux_style)
print(windows_style)
```

```python
# Concrete Path Classes (Filesystem Interaction)
from pathlib import Path

real_path = Path("projects/data")
print(real_path.exists())
```

```python
# Path, PosixPath, WindowsPath (Auto Selection)
from pathlib import Path

current_path = Path("sample_folder")
print(type(current_path))
```

```python
# Getting Current Working Directory
from pathlib import Path

current_dir = Path.cwd()
print(current_dir)
```

```python
# Path Existence and Validation
from pathlib import Path

check_path = Path("projects/config/settings.yaml")

if check_path.exists():
    print("Path is valid")
else:
    print("Path does not exist")
```

```python
# Directory and File Type Checks
from pathlib import Path

item = Path("projects/data")

print("Is file:", item.is_file())
print("Is directory:", item.is_dir())
```
