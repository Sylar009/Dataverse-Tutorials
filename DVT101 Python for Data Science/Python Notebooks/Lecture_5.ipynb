```python
#Arithmetic Operators
num_one = 15
num_two = 4

print("Addition Result:", num_one + num_two)
print("Subtraction Result:", num_one - num_two)
print("Multiplication Result:", num_one * num_two)
print("Division Result:", num_one / num_two)
print("Floor Division Result:", num_one // num_two)
print("Modulus Result:", num_one % num_two)
print("Power Result:", num_one ** num_two)
```

```python
# Comparison Operators
value_a = 13
value_b = 33

print(value_a > value_b)
print(value_a < value_b)
print(value_a == value_b)
print(value_a != value_b)
print(value_a >= value_b)
print(value_a <= value_b)
```

```python
# Logical Operators
flag_one = True
flag_two = False

print(flag_one and flag_two)
print(flag_one or flag_two)
print(not flag_one)
```

```python
# Bitwise Operators
bit_x = 10
bit_y = 4

print(bit_x & bit_y)
print(bit_x | bit_y)
print(bit_x ^ bit_y)
print(~bit_x)
print(bit_x >> 2)
print(bit_x << 2)
```

```python
# Assignment Operators
score = 10
score += 5
score -= 3
score *= 2
print(score)
```

```python
# Ternary Operator
num_a = 10
num_b = 20

smallest = num_a if num_a < num_b else num_b
print(smallest)
```
