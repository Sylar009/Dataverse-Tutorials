```python
# Directory Management Using pathlib
from pathlib import Path

folder = Path("analysis_reports")
folder.mkdir(exist_ok=True)

for entry in Path(".").iterdir():
    print(entry)
```

```python
# Calculating Directory Size
import os

total_bytes = 0

for root_path, sub_dirs, files in os.walk(os.getcwd()):
    for file_name in files:
        full_path = os.path.join(root_path, file_name)
        total_bytes += os.path.getsize(full_path)

print("Directory size:", total_bytes, "bytes")
```

```python
# Access and Modification Time
import os
import time

target_path = "/"

access_ts = os.path.getatime(target_path)
modify_ts = os.path.getmtime(target_path)

print("Last Access:", time.ctime(access_ts))
print("Last Modified:", time.ctime(modify_ts))
```
