```python
# Numeric Data Types
whole_num = 12
decimal_num = 12.75
complex_num = 3 + 5j

print(type(whole_num))
print(type(decimal_num))
print(type(complex_num))
```

```python
# String Data Type
message = "Welcome to Dataverse Tutorials"

print(message)
print(type(message))
print(message[0])
print(message[-1])
```

```python
# List Data Type
student_data = ["Aman", 22, 85.5, True]

print(student_data)
print(student_data[1])
```

```python
# Tuple Data Type
coordinates = (10.5, 20.8)

print(coordinates)
print(coordinates[0])
```

```python
# Boolean Data Type
is_logged_in = True
print(type(is_logged_in))
```

```python
# Set Data Type
unique_ids = {101, 102, 101, 103}

print(unique_ids)
```

```python
# Dictionary Data Type
employee = {
    "name": "Riya",
    "age": 28,
    "role": "Data Analyst"
}

print(employee["name"])
print(employee.get("role"))
```
