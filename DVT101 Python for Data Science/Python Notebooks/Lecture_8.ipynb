```python
# If Statement (Basic Level)
user_age = 20

if user_age >= 18:
    print("You are eligible to vote")
```

```python
# If-Else Statement
ticket_age = 10

if ticket_age <= 12:
    print("Free entry allowed")
else:
    print("Ticket payment required")
```

```python
# Elif Statement (Multiple Conditions)
person_age = 25

if person_age <= 12:
    print("Child")
elif person_age <= 19:
    print("Teenager")
elif person_age <= 35:
    print("Young Adult")
else:
    print("Adult")
```

```python
# Nested If-Else
age_value = 70
premium_member = True

if age_value >= 60:
    if premium_member:
        print("30% discount applied")
    else:
        print("20% discount applied")
else:
    print("No discount available")
```

```python
# Ternary Conditional Statement
age_check = 20
status_label = "Adult" if age_check >= 18 else "Minor"
print(status_label)
```

```python
# Match-Case Statement (Advanced)
input_number = 2

match input_number:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("Other Number")
```
