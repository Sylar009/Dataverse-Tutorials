```python
# For Loop – Basic Example
max_limit = 4

for current_number in range(max_limit):
    print(current_number)
```

```python
# For Loop with Collections
topics = ["Python", "Data", "Machine Learning"]

for subject in topics:
    print(subject)
```

```python
# While Loop – Condition Based Loop
counter = 0

while counter < 3:
    print("Learning Python")
    counter += 1
```

```python
# Nested Loops – Loop inside Loop
for row in range(1, 5):
    for col in range(row):
        print(row, end=" ")
    print()
```

```python
# Loop Control Statements – break & continue
for letter in "dataverse":
    if letter == "a":
        continue
    print(letter)
```

```python
for letter in "python":
    if letter == "h":
        break
    print(letter)
```
